```python
#asks user for 2 numbers for the operation
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

#addition function
def add_numbers(num1,num2):
  sum = num1 + num2
  return sum

#subtraction function
def subtract_numbers(num1,num2):
  difference = num1 - num2
  return difference

#multiplication function
def multiply_numbers(num1,num2):
  product = num1 * num2
  return product

#division function
def divide_numbers(num1,num2):
  quotient = num1 / num2
  return quotient

 #asks user for the desired operatiom
option = int(input("1 is addition,2 is subtraction,3 is multiplication,4 is division.Enter Desired Operation: "))

#picks an operation based on what the user inputed
if option == 1:
 total = add_numbers(num1,num2)
 print(total)
elif option == 2:
 total = subract_numbers(num1,num2)
 print(total)
elif option == 3:
 total = multiply_numbers(num1,num2)
 print(total)
elif option == 4:
 total = divide_numbers(num1,num2)
 print(total)


```

2.All my functions used the parameters num1 and num2.

3.The arguments that passed through the functions is the 2 numbers that the user typed.For example if the 2 numbers were 2 and 4 then that would be the argument.
